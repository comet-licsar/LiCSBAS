```python
# How to add viewing functionality to ipynb:
# - install xcube - e.g. mamba install -c conda-forge xcube
# - then export your cum*h5 file to netcdf with --xcube, e.g.:
# LiCSBAS_out2nc.py -i TS*/cum.h5 -o exported.nc --xcube
# - then load it into the notebook, e.g.
import xarray as xr
ds = xr.open_dataset('exported.nc')
```

```python
from xcube.webapi.viewer import Viewer
viewer.add_dataset(ds)
```

```python
viewer.info()
```

```python
viewer.show()
```

```python
# if the above did not work, the way is to set manually the style - but.. the above should work...
viewer = Viewer(
    server_config={
        "Styles": [
            {
                "Identifier": "SST",
                "ColorMappings": {
                    "analysed_sst": {"ValueRange": [270, 300], "ColorBar": "inferno"}
                },
            }
        ]
    }
)


viewer.add_dataset(ds, style="SST")
```
